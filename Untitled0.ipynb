```python
class BankAccount:
  def __init__(self, holder, balance):
    self.holder = holder
    self.balance = balance

  def deposit(self, amount):
    self.balance += amount

  def withdraw(self, amount):
    if amount > self.balance:
      print("Insufficient balance")
    else:
      self.balance -= amount
      print("withdraw successful")

  def show_balance(self):
    print(self.balance)

bank_acc1 = BankAccount("Arvind", 1000)
bank_acc1.deposit(500)
bank_acc1.show_balance()
bank_acc1.withdraw(2000)


```

# else block runs only when no exception occurs

```python
try:
  num = int(input())
except ValueError:
  print("Invalid input")
else:
  print("You entered:", num)
```

```python
#finally block always executes irrespective whether there is an error or not
try:
  num = int(input())
except ValueError:
  print("Invalid input")
finally:
  print("program finished")
```
